```python
import matplotlib.pyplot as plt
import numpy as np

models = ['DT Default', 'DT Tuned', 'RF Default', 'RF Tuned']

r2_scores = [0.7323, 0.7458, 0.8557, 0.8598]
mae_scores = [30.8056, 30.1222, 24.4472, 24.1644]

fig, ax = plt.subplots(1, 2, figsize=(16, 6))

colors = ['#a6cee3', '#1f78b4', '#b2df8a', '#33a02c']

bars1 = ax[0].bar(models, r2_scores, color=colors, edgecolor='black', alpha=0.9)
ax[0].set_title('Comparison of R2 Score (Higher is Better)', fontsize=14, fontweight='bold')
ax[0].set_ylabel('R2 Score', fontsize=12)
ax[0].set_ylim(0.6, 1.0)
ax[0].grid(axis='y', linestyle='--', alpha=0.7)

for bar in bars1:
    height = bar.get_height()
    ax[0].text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.4f}', ha='center', va='bottom', fontsize=11)

bars2 = ax[1].bar(models, mae_scores, color=colors, edgecolor='black', alpha=0.9)
ax[1].set_title('Comparison of MAE (Lower is Better)', fontsize=14, fontweight='bold')
ax[1].set_ylabel('Mean Absolute Error (Tensile Strength)', fontsize=12)
ax[1].set_ylim(0, 35)
ax[1].grid(axis='y', linestyle='--', alpha=0.7)

for bar in bars2:
    height = bar.get_height()
    ax[1].text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.2f}', ha='center', va='bottom', fontsize=11)

plt.suptitle('Algorithm Performance Comparison: Decision Tree vs Random Forest', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv('steel.csv')

plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Figure 4: Feature Correlation Matrix')
plt.show()

y = df['tensile_strength']
y_train, y_test = train_test_split(y, test_size=0.2, random_state=42)

plt.figure(figsize=(10, 6))
sns.kdeplot(y_train, label='Training Set', fill=True, color='blue', alpha=0.3)
sns.kdeplot(y_test, label='Test Set', fill=True, color='red', alpha=0.3)
plt.title('Figure 6: Target Distribution after Train-Test Split')
plt.xlabel('Tensile Strength')
plt.legend()
plt.show()
```
